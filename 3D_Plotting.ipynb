```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df1 = pd.read_csv("./ev_battery_failure_dataset.csv").copy().reset_index(drop = True)
df1
```

```python
fig = plt.figure()
ax = plt.axes(projection='3d')

x = df1['sensor_fault_count']
y = df1['BMS_warning_count']
z = df1['battery_failure']
c = x + y + z

fig = plt.figure(figsize=(100, 100))

ax.scatter(x, y, z, c = c)

ax.set_title('3D Scatter')
ax.set_xlabel('sensor_fault_count')
ax.set_ylabel('BMS_warning_count')
ax.set_zlabel('battery_failure')

plt.show()
```
