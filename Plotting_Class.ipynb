```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
df1 = pd.read_csv("./ev_battery_failure_dataset.csv").copy().reset_index(drop = True)
df1
```

```python
class plotting(object):

    def __init__(self, dataset):
        self.dataset = dataset.copy().reset_index(drop = True)

    def plotting_function(self):
        dataset_plotting = self.dataset.copy().drop(columns = ['vehicle_id', 'battery_serial'])
        column_list = list(dataset_plotting)
        column_list_len = len(column_list)-1
        for i in range(0, column_list_len):
            sns.scatterplot(x=dataset_plotting.iloc[:, i], y=dataset_plotting["battery_failure"])
            plt.show()
            
    def main(self):
        self.plotting_function()
```

```python
plotting = plotting(df1)
```

```python
plotting.main()
```
