```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df1 = pd.read_csv("./ev_battery_failure_dataset.csv").copy().reset_index(drop = True)
df1
```

```python
class ThreeDplotting(object):

    def __init__(self, dataset):
        self.dataset = dataset.copy().reset_index(drop = True)

    def Num_Dataset(self):
        global dataset_num_global
        numerics = ['int16', 'int32', 'int64', 'float16', 'float32', 'float64']

        dataset_num = self.dataset.copy().reset_index(drop = True)
        dataset_num_global = dataset_num.select_dtypes(include=numerics)

    def ThreeDplotting_function(self):
        column_list = list(dataset_num_global)
        column_list_len = len(column_list)-1
        for i in range(46, 47):
            for j in range(1, column_list_len):
                if (i != j):
                    fig = plt.figure()
                    ax = plt.axes(projection='3d')

                    x = dataset_num_global.iloc[:, i]
                    y = dataset_num_global.iloc[:, j]
                    z = df1['battery_failure']
                    c = x + y + z

                    fig = plt.figure(figsize=(100, 100))

                    ax.scatter(x, y, z, c = c)

                    ax.set_title('3D Scatter')
                    ax.set_xlabel(column_list[i])
                    ax.set_ylabel(column_list[j])
                    ax.set_zlabel('battery_failure')

                    plt.show()
            
    def main(self):
        self.Num_Dataset()
        self.ThreeDplotting_function()
```

```python
ThreeDplotting = ThreeDplotting(df1)
```

```python
ThreeDplotting.main()
```
